1.What is Boosting in Machine Learning? Explain how it improves weak laerner?
  Boosting in Machine Leraning refers to process of trainnig multiple models sequentially that are dependent on each other.It improves weak learners by learning the mistakes from previous rectifying them in further models.

2.What is the difference between AdaBoost and Gradient Boosting in terms of how models are trained?
  Ada Boost- It increases the weight of misclassified data points in further models to reduce the bias at an overall level.
  Gradient Boost-It uses the resdidual error for each data point for training the models.

3.How does regularization help in XGBoost?
  Regularization helps in XGBoost by reducing the variance thus helping in avoiding the overfitting.It makes trees simplifis by enalizing large leaf weights.

4.Why is CatBoost considered efficient for handling categorical data?
  CatBoost is considered effiecnet for categorical data as we do not have to manually convert the categorical values to numerical it itself does that by uilizing ordered target guided encoding.

5.What are some real-world applications where boosting techniques are preferred over bagging methods?
  fraud detection, financial forecasting, customer churn prediction, and medical diagnosis.

```python
# 6.Train an AdaBoost Classifier on the Breast Cancer dataset
# ● Print the model accuracy
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()
x=data.data
y=data.target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
model=AdaBoostClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,model.predict(x_test))
```

```python
# 8. Write a Python program to:
# ● Train an XGBoost Classifier on the Breast Cancer dataset
# ● Tune the learning rate using GridSearchCV
# ● Print the best parameters and accuracy
from sklearn.datasets import load_breast_cancer
data=load_breast_cancer()
x=data.data
y=data.target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from xgboost import XGBClassifier
model=XGBClassifier()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
params={
    "n_estimators": [1,2,1,20,30,50,100],
    "learning_rate": [0.1,0.2,0.3,0.4]
}
```

```python
clf=GridSearchCV(model,param_grid=params,verbose=3,cv=5)
```

```python
clf.fit(x_train,y_train)
```

```python
clf.best_params_
```

```python
clf=clf.best_estimator_
```

```python
clf.predict(x_test)
```

```python
accuracy_score(y_test,clf.predict(x_test))
```
